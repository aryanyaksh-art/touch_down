# 08 · Website replays on Kaggle (no Drive needed)

Re-flies a handful of chosen seeds with camera frames recorded and writes the website replay data to `/kaggle/working/web_data.zip`. Seeds and titles come from `python scripts/pick_replay_seeds.py DIR` run on the finished Monte Carlo output (set `SEEDS` and `TITLES` below to what it prints).

Settings: **GPU T4 x2**, **Internet on**, add the private dataset `touchdown-detector-v2` as an input. Replays are re-flown on a different GPU from the Monte Carlo, so tiny rendering differences are possible; the website says so.


```python
!git clone -q https://github.com/aryanyaksh-art/touch_down.git /tmp/touch_down || (cd /tmp/touch_down && git pull -q)
!sh /tmp/touch_down/scripts/kaggle_setup.sh 2>&1 | tail -5
```

```python
import os, shlex
# the v2 detector was uploaded as two parts; join them back
!find /kaggle/input -name "best.pt.part*" | sort | xargs cat > /tmp/best.pt
CKPT = '/tmp/best.pt'
assert os.path.getsize(CKPT) == 17520863, os.path.getsize(CKPT)
# paste the output of scripts/pick_replay_seeds.py here
SEEDS = [0, 1, 2]
TITLES = ['Clean touchdown', 'Justified back-away', 'Typical descent']
```

```python
cmd = ('cd /tmp/touch_down && TOUCHDOWN_DATA=/tmp/data TOUCHDOWN_BLENDER=/tmp/blender/blender '
       'python -m touchdown.analysis.export_replays --out /kaggle/working/web_data --device GPU --aim-mode random_safe '
       f'--checkpoint {CKPT} --seeds ' + ' '.join(map(str, SEEDS)) + ' --titles ' + ' '.join(shlex.quote(t) for t in TITLES))
!{cmd}
!cd /kaggle/working && zip -qr web_data.zip web_data && ls -la web_data.zip && du -sh web_data
```
